```python
import numpy as np
```

```python
chartspace_range = np.array([113.76, 650.61])
logspace_range = np.array([2, 6])
points = np.array(
    [
        116.72,
        240.59,
        322.88,
        364.16,
        405.48,
        446.73,
        488.03,
        529.83,
        570.58,
        611.62,
        652.91,
    ]
)
points_offsetted = points - ((points[0] - chartspace_range[0]))
```

```python
# Map points to 2 to 6 based on chartspace_range
points_normalized = (points_offsetted - chartspace_range[0]) / (
    chartspace_range[1] - chartspace_range[0]
)

points_normalized = (
    points_normalized * (logspace_range[1] - logspace_range[0])
) + logspace_range[0]
```

```python
(10 ** points_normalized).astype(int).tolist()
```
